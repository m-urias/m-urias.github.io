# Melitz Model of Intra-Industry Trade


## Motivation

The Krugman model establishes that internal economies of scale generate gains from trade even between identical countries, but it does so with a strong symmetry assumption: all firms are equally productive, so trade affects the scale of production but not the composition of firms. This sits uneasily with a robust body of empirical evidence documenting that firms within the same industry differ dramatically in their productivity, and that trade systematically reallocates activity toward the most productive among them.

Several empirical regularities motivate a richer framework:

- Exporters are systematically larger, more productive, more capital-intensive, and pay  higher wages than non-exporters within the same industry.
- Only a minority of firms export — and those that do tend to be drawn from the upper tail  of the productivity distribution.
- Trade liberalization is associated with significant entry and exit: less productive firms  contract or exit, while more productive firms expand, generating aggregate productivity  gains through reallocation rather than through improvements at any individual firm.

Bernard and Jensen (1999) quantify this last effect for US manufacturing, finding that 
market share reallocations toward more productive exporting plants account for approximately 20% of US manufacturing productivity growth. In the Canadian context, Trefler (2004) finds that about half of the long-run productivity gains from the Canada-US FTA are attributable to plant turnover — the exit of less productive plants and their replacement by more productive entrants. These selection effects are absent by construction from the symmetric Krugman model.

Melitz (2003) extends the Krugman framework to incorporate firm-level productivity 
heterogeneity in a general equilibrium setting. The model endogenizes which firms produce, which export, and which exit, generating a rich set of predictions about the within-industry reallocation effects of trade. The key ingredients are: (i) monopolistic competition with CES preferences, (ii) firm-level productivity draws from a common distribution, and (iii) a forward-looking entry decision in which firms pay a sunk entry cost before observing their productivity. We develop these in turn.


## Environment

### Preferences and Demand

There is a continuum of consumers with identical CES preferences over a continuum of 
differentiated varieties $\omega \in \Omega$:

$$U = \left[\int_{\omega \in \Omega} q(\omega)^{\frac{\sigma-1}{\sigma}} d\omega \right]^{\frac{\sigma}{\sigma-1}}$$

where $\sigma > 1$ is the elasticity of substitution between any two varieties. Let 
$\rho \equiv (\sigma-1)/\sigma \in (0,1)$ so that $U = \left[\int_\Omega q(\omega)^\rho 
d\omega\right]^{1/\rho}$. Each consumer maximizes utility subject to the budget constraint:

$$\int_{\omega \in \Omega} p(\omega) q(\omega) d\omega \leq I$$

### Deriving the Demand Function

To derive demand, it is convenient to work with the dual problem: minimize the expenditure required to achieve utility level $U$,

$$\min_{q(\omega)} \int_{\omega \in \Omega} p(\omega) q(\omega) d\omega \quad \text{s.t.} 
\quad \left[\int_{\omega \in \Omega} q(\omega)^\rho d\omega\right]^{1/\rho} = U$$

The first-order condition equates the marginal rate of substitution between any two 
varieties $i$ and $j$ to their price ratio:

$$\frac{q(i)^{\rho-1}}{q(j)^{\rho-1}} = \frac{p(i)}{p(j)}$$

Solving for $q(i)$ in terms of $q(j)$ and substituting into the constraint yields the 
compensated (Hicksian) demand for variety $j$:

$$q(j) = \frac{p(j)^{\frac{1}{\rho-1}}}{\left[\int_{\omega \in \Omega} 
p(i)^{\frac{\rho}{\rho-1}} di\right]^{1/\rho}} U$$

The minimum cost of purchasing a bundle that delivers one unit of utility — the **aggregate price index** — is obtained by substituting the compensated demands back into the 
expenditure function:

$$\int_{\omega \in \Omega} p(j) q(j) dj = \underbrace{\left[\int_{\omega \in \Omega} 
p(i)^{\frac{\rho}{\rho-1}} di\right]^{\frac{\rho-1}{\rho}}}_{\equiv\; P} \cdot U$$

The price index $P$ measures the minimum cost of one unit of the aggregate consumption 
bundle. Expressing $\frac{\rho}{\rho-1} = 1-\sigma$ and $\frac{\rho-1}{\rho} = 
\frac{1}{1-\sigma}$, we can write $P$ in its standard form:

$$P = \left[\int_{\omega \in \Omega} p(\omega)^{1-\sigma} d\omega\right]^{\frac{1}{1-\sigma}}$$

Demand for variety $\omega$ can now be written compactly in terms of $P$:

$$q(\omega) = \left(\frac{p(\omega)}{P}\right)^{-\sigma} Q$$

where $Q = R/P$ is aggregate real consumption and $R = \int_\Omega r(\omega) d\omega$ is aggregate nominal expenditure. The implied revenue earned by the producer of variety 
$\omega$ is:

$$r(\omega) = p(\omega) q(\omega) = R \left(\frac{p(\omega)}{P}\right)^{1-\sigma}$$

Several properties of this demand system are worth noting. First, the demand elasticity facing each firm is constant and equal to $\sigma$, regardless of the prices charged by other firms — a consequence of the CES structure that will greatly simplify the firm's pricing problem. Second, revenue is decreasing in a firm's own price relative to the aggregate price index: a firm that charges a higher price captures a smaller share of total expenditure $R$. Third, the CES structure implies that varieties enter symmetrically — consumers have love of variety in the sense that, holding the price index $P$ constant, welfare is increasing in the mass of available varieties.

### Welfare and the Price Index

Substituting the demand functions back into the utility function confirms that equilibrium welfare can be expressed entirely in terms of the price index:

$$\begin{align*}
U &= \left[\int_{\omega \in \Omega} \left(\frac{p(\omega)}{P}\right)^{1-\sigma} 
Q^{\frac{\sigma-1}{\sigma}} d\omega\right]^{\frac{\sigma}{\sigma-1}} \\
&= Q \cdot P^\sigma \left[\int_{\omega \in \Omega} p(\omega)^{1-\sigma} 
d\omega\right]^{\frac{\sigma}{\sigma-1}} \\
&= Q \cdot P^\sigma \cdot P^{-\sigma} \cdot P \\
&= \frac{I}{P}
\end{align*}$$

With income normalized to $I = 1$, welfare reduces to:

$$U = \frac{1}{P}$$

This is a powerful simplification: **welfare is completely summarized by the aggregate 
price index.** Any change in market structure, technology, or trade policy that lowers $P$ raises welfare, and the magnitude of the welfare gain is fully captured by the percentage decline in $P$. This result will allow us to characterize the welfare effects of trade liberalization directly from its impact on the price index — which in turn depends on the mass of available varieties and their individual prices, both of which the Melitz model endogenizes.


### Production Technology and Firm Behavior

#### Technology

Labor is the only factor of production, with the total labor supply fixed at $L$ and the wage normalized to unity: $w = 1$. Each variety is produced by a single firm. The total cost of producing $q$ units for a firm with productivity $\varphi$ is:

$$TC(\varphi) = f + \frac{q(\varphi)}{\varphi}$$

where $f > 0$ is a fixed operating cost (incurred regardless of output) and $1/\varphi$ is the marginal cost of production. Since $w = 1$, this expression simultaneously represents labor demand and total cost. The key feature is that more productive firms — those with higher $\varphi$ — have lower marginal costs, and therefore can profitably undercut less productive rivals on price. This is the source of firm heterogeneity in the model.

The production side differs from Krugman in one critical respect: rather than assuming all firms share the same productivity, each firm draws its productivity $\varphi$ from a common distribution $G(\varphi)$ after paying a sunk entry cost $f_e$. Each active firm also faces an exogenous probability of death $\delta$ each period — a reduced-form representation of firm exit due to demand shocks, obsolescence, or managerial failure. We return to the entry decision below; for now, we take $\varphi$ as given and characterize how firm behavior varies with productivity.

#### Optimal Pricing

A firm with productivity $\varphi$ chooses its price to maximize profits, taking the 
aggregate price index $P$ and total expenditure $R$ as given. Profits are:

$$\pi(\varphi) = r(\varphi) - TC(\varphi) = p(\varphi)q(\varphi) - \frac{q(\varphi)}{\varphi} - f$$

Substituting the demand function $q(\varphi) = (p(\varphi)/P)^{-\sigma}Q$ and taking the first-order condition with respect to $p(\varphi)$:

$$p(\varphi) = \frac{\sigma}{\sigma - 1} \cdot \frac{1}{\varphi} = \frac{1}{\rho\varphi}$$

This is the standard CES markup pricing rule: each firm charges a constant markup 
$\sigma/(\sigma-1) > 1$ over its marginal cost $1/\varphi$. Since the markup is identical across firms, price differences across firms reflect only differences in marginal cost — more productive firms charge lower prices and therefore capture larger market shares.

#### Firm-Level Output, Revenue, and Profit

Substituting the optimal price back into the demand and revenue expressions:

$$\begin{align*}
q(\varphi) &= RP^{\sigma-1}(\rho\varphi)^\sigma \\
r(\varphi) &= R(P\rho\varphi)^{\sigma-1} \\
\pi(\varphi) &= \frac{r(\varphi)}{\sigma} - f
\end{align*}$$

The profit expression $\pi(\varphi) = r(\varphi)/\sigma - f$ has a clean interpretation: a fraction $1/\sigma$ of revenue covers variable costs net of the markup, and the remainder must cover the fixed cost $f$. Since $r'(\varphi) > 0$, both revenue and profit are strictly increasing in productivity. This monotonicity is central to the model's predictions: it means that more productive firms are larger, earn higher profits, and — as we will see — are the ones that survive and export.

### Equilibrium

#### The Productivity Cutoff

After drawing productivity $\varphi$, a firm decides whether to produce or exit. The 
expected discounted value of profits for a firm that produces is:

$$\nu(\varphi) = \max\left\{0,\; \sum_{t=s}^{\infty}(1-\delta)^{t-s}\pi(\varphi)\right\} 
= \max\left\{0,\; \frac{\pi(\varphi)}{\delta}\right\}$$

where the second equality uses the formula for a geometric series. Since $\pi(\varphi)$ is strictly increasing in $\varphi$, there exists a unique **productivity cutoff** $\varphi^*$ below which a firm prefers to exit rather than produce:

$$\nu(\varphi) > 0 \iff \varphi > \varphi^*$$

The cutoff $\varphi^*$ is pinned down by the **zero cutoff profit (ZCP) condition** — the requirement that the marginal active firm earns exactly zero profit:

$$\pi(\varphi^*) = 0 \qquad \Longleftrightarrow \qquad r(\varphi^*) = \sigma f$$

Firms with $\varphi < \varphi^*$ exit immediately upon observing their draw; firms with $\varphi \geq \varphi^*$ remain active. The cutoff $\varphi^*$ is an endogenous object that will be determined in equilibrium.

#### The Distribution of Active Firms

Let $M$ denote the equilibrium mass of active firms. Conditional on survival, the 
productivity of an active firm is drawn from the truncated distribution:

$$\mu(\varphi) = \begin{cases} \dfrac{g(\varphi)}{1 - G(\varphi^*)} & \text{if } \varphi \geq \varphi^* \\[6pt] 0 & \text{if } \varphi < \varphi^* \end{cases}$$

This is the ex-post productivity distribution — the distribution of productivity among 
firms that have survived the entry selection. As $\varphi^*$ rises, the truncation becomes more severe and the surviving pool of firms is more productive on average. This is the selection mechanism at the heart of the model.

#### Aggregation and the Average Productivity

Melitz introduces an aggregation device that greatly simplifies the analysis. Define the **average productivity** of active firms as the productivity-weighted average:

$$\tilde{\varphi}(\varphi^*) = \left[\frac{1}{1-G(\varphi^*)}\int_{\varphi^*}^{\infty} 
\varphi^{\sigma-1} g(\varphi)\, d\varphi\right]^{\frac{1}{\sigma-1}}$$

The weights $\varphi^{\sigma-1}$ reflect the fact that more productive firms contribute disproportionately to aggregate revenue and welfare under CES preferences — a firm with productivity twice as high generates $2^{\sigma-1}$ times the revenue. The average $\tilde{\varphi}$ is a sufficient statistic for the productivity distribution of active firms: it is uniquely determined by $\varphi^*$ and the exogenous distribution $G(\varphi)$, so once we know $\varphi^*$ we know $\tilde{\varphi}$.

The value of this aggregation device becomes clear when we compute aggregate variables. The price index across $M$ active firms is:

$$P = \left[M\int_{\varphi^*}^{\infty} p(\varphi)^{1-\sigma}\mu(\varphi)\, 
d\varphi\right]^{\frac{1}{1-\sigma}}$$

Substituting the optimal price $p(\varphi) = 1/\rho\varphi$:

$$P = \frac{M^{\frac{1}{1-\sigma}}}{\rho}\left[\int_{\varphi^*}^{\infty} 
\varphi^{\sigma-1}\mu(\varphi)\, d\varphi\right]^{\frac{1}{1-\sigma}} 
= \frac{M^{\frac{1}{1-\sigma}}}{\rho\tilde{\varphi}}$$

The price index depends on only two objects: the mass of active varieties $M$, which 
captures the love-of-variety effect ($P$ falls as $M$ rises), and the average productivity $\tilde{\varphi}$, which captures the efficiency of active producers ($P$ falls as $\tilde{\varphi}$ rises). The same aggregation yields expressions for aggregate revenue, profit, and output in terms of the representative firm with productivity $\tilde{\varphi}$:

$$R = Mr(\tilde{\varphi}), \qquad \Pi = M\pi(\tilde{\varphi}), \qquad 
Q = M^{\frac{\sigma}{\sigma-1}}q(\tilde{\varphi})$$

The economy therefore behaves as if it were populated by $M$ identical firms, each with productivity $\tilde{\varphi}$ — an exact aggregation result that reduces the 
heterogeneous-firm economy to a representative-firm problem without losing any information.

#### The Zero Cutoff Profit Condition

We now derive a key equilibrium relationship. Average profit can be written as:

$$\bar{\pi} = \frac{r(\tilde{\varphi})}{\sigma} - f$$

Since revenue scales with productivity as $r(\varphi) \propto \varphi^{\sigma-1}$, we have:

$$r(\tilde{\varphi}) = r(\varphi^*)\left(\frac{\tilde{\varphi}}{\varphi^*}\right)^{\sigma-1}$$

Applying the ZCP condition $r(\varphi^*) = \sigma f$:

$$\bar{\pi} = \frac{r(\tilde{\varphi})}{\sigma} - f 
= f\left[\left(\frac{\tilde{\varphi}(\varphi^*)}{\varphi^*}\right)^{\sigma-1} - 1\right]$$

This expression — the **ZCP condition** — relates average profit to the productivity 
cutoff $\varphi^*$ through the ratio $\tilde{\varphi}(\varphi^*)/\varphi^*$. Several 
properties are worth noting. First, $\bar{\pi} \geq 0$ always, since $\tilde{\varphi} 
\geq \varphi^*$ by construction (the average of a distribution truncated at $\varphi^*$ cannot be below $\varphi^*$). Second, $\bar{\pi}$ is increasing in $\varphi^*$: a higher cutoff means the surviving pool of firms is more productive on average relative to the marginal firm, so average profits are higher. This monotonicity will be essential when we close the model with the free entry condition in the next section.

### Closing the Model: The Free Entry Condition

#### The Zero Cutoff Profit Condition

We established in the previous section that average profit satisfies:

$$\bar{\pi} = f\left[\left(\frac{\tilde{\varphi}(\varphi^*)}{\varphi^*}\right)^{\sigma-1} - 1\right]$$

The relationship between $\bar{\pi}$ and $\varphi^*$ encoded in this expression is not 
immediately obvious, because both $\tilde{\varphi}$ and $\varphi^*$ appear. To determine the slope of the ZCP locus, note that:

$$\tilde{\varphi}'(\varphi^*) = \frac{g(\varphi^*)\tilde{\varphi}^{2-\sigma}}{(\sigma-1)^{2}}
\int_{\varphi^*}^{\infty}(\varphi^{\sigma-1} - \varphi^{*\sigma-1})g(\varphi)\, d\varphi > 0$$

so $\tilde{\varphi}$ is increasing in $\varphi^*$. The numerator $\tilde{\varphi}(\varphi^*)^{\sigma-1}$ and the denominator $\varphi^{*\sigma-1}$ in the ZCP expression are therefore both increasing in $\varphi^*$, and their ratio — and hence 
$\bar{\pi}$ — could in principle be increasing or decreasing.

The two forces at work reflect a tension inherent in the model. On one hand, a higher 
cutoff $\varphi^*$ directly raises the average productivity of surviving firms, which tends to increase average profits. On the other hand, a higher cutoff also means that surviving firms face more productive rivals, which intensifies competition and tends to reduce profits. Whether $\bar{\pi}$ rises or falls with $\varphi^*$ depends on which force dominates, which in turn depends on the shape of the productivity distribution $G(\varphi)$. If $G$ has a fat enough right tail, competition effects dominate and the ZCP locus is downward-sloping. For the Pareto distribution — the most commonly used parametrization in the literature — the two effects exactly offset, and the ZCP locus is flat. More generally, the ZCP is downward-sloping, and we proceed under this assumption.

#### The Free Entry Condition

The second equilibrium condition comes from the entry decision. Before observing their 
productivity, potential entrants pay a sunk cost $f_e$ to enter. Upon observing their draw $\varphi \sim G(\varphi)$, they produce if $\varphi \geq \varphi^*$ and exit otherwise. In a stationary equilibrium, free entry drives the expected value of entry to zero:

$$\underbrace{(1-G(\varphi^*))\frac{\bar{\pi}}{\delta}}_{\text{expected profit from entry}} 
= \underbrace{f_e}_{\text{sunk entry cost}}$$

Rearranging, the **free entry (FE) condition** is:

$$\bar{\pi} = \frac{\delta f_e}{1 - G(\varphi^*)}$$

The FE condition says that average profit must be just sufficient to compensate entrants for the probability of drawing a productivity below the cutoff and being forced to exit. A higher cutoff $\varphi^*$ means a lower survival probability $1 - G(\varphi^*)$, so surviving firms must earn higher average profits to justify the sunk entry cost. The FE locus is therefore **upward-sloping** in $(\varphi^*, \bar{\pi})$ space.

#### Equilibrium Determination

The equilibrium $(\varphi^*, \bar{\pi})$ is determined at the intersection of the ZCP and FE loci:

Since the ZCP is downward-sloping and FE is upward-sloping, the intersection is unique 
under standard regularity conditions. Once $(\varphi^*, \bar{\pi})$ are determined, all remaining equilibrium objects follow. From labor market clearing — $L = R = r(\tilde{\varphi})M$ — the equilibrium mass of active firms is:

$$M = \frac{L}{r(\tilde{\varphi})} = \frac{L}{\sigma(\bar{\pi} + f)}$$

Since $\bar{\pi}$ is independent of $L$ (it is determined entirely by the ZCP and FE 
conditions, which do not involve $L$), the mass of varieties is proportional to the size of the economy. A larger economy supports more firms and hence more varieties — the same result as in Krugman, now emerging from a heterogeneous-firm environment. Welfare of the representative consumer is:

$$U = \frac{1}{P} = M^{\frac{1}{\sigma-1}}\rho\tilde{\varphi}$$

Welfare depends on two sufficient statistics: the mass of varieties $M$ (love of variety) and average productivity $\tilde{\varphi}$ (productive efficiency). An increase in $L$ raises $M$ proportionally, raising welfare through the variety channel — exactly as in Krugman. The novel feature of the Melitz model is that $\tilde{\varphi}$ can also vary, providing an additional productivity channel for welfare gains, which becomes active when the economy opens to trade.

A final observation: once $\tilde{\varphi}$ and $\bar{\pi}$ are known, the aggregate 
outcome of this heterogeneous-firm economy is exactly replicated by a representative firm with productivity $\tilde{\varphi}$ and profit $\bar{\pi}$. The aggregation is exact — no information about the distribution of firm-level outcomes is lost in passing to the representative firm — which is what makes the model tractable despite the full distribution of heterogeneity.

### Opening to Trade

#### Trade Costs

We now introduce two types of trade frictions that create a distinction between domestic and export activity:

- **Iceberg trade costs**: shipping one unit of a good to a foreign market requires 
  producing $\tau \geq 1$ units, so a fraction $1 - 1/\tau$ melts away in transit. The 
  marginal cost of delivering one unit abroad is therefore $\tau/\varphi$, versus $1/\varphi$ domestically.
- **Fixed export costs**: to access a foreign market, a firm must pay a per-period fixed cost $f_x$ (the amortized value of the sunk export entry cost $f_{ex} = f_x/\delta$). 
  This captures one-time investments in distribution networks, regulatory compliance, and market information that must be incurred before exporting can begin.

We consider $n$ symmetric foreign countries, with wages normalized to unity in all 
countries. Monopoly pricing now yields separate prices for domestic and export sales:

$$p_d(\varphi) = \frac{1}{\rho\varphi}, \qquad p_x(\varphi) = \frac{\tau}{\rho\varphi}$$

and corresponding revenues:

$$r_d(\varphi) = R(P\rho\varphi)^{\sigma-1}, \qquad 
r_x(\varphi) = \tau^{1-\sigma}R(P\rho\varphi)^{\sigma-1}$$

where symmetry implies $P_d = P_x = P$ and $R_d = R_x = R$ across markets. Note that 
$r_x(\varphi) = \tau^{1-\sigma}r_d(\varphi) < r_d(\varphi)$: iceberg costs reduce export revenue relative to domestic revenue by the factor $\tau^{1-\sigma} < 1$.

Profits from domestic sales and exporting are:

$$\pi_d(\varphi) = \frac{r_d(\varphi)}{\sigma} - f, \qquad 
\pi_x(\varphi) = \frac{r_x(\varphi)}{\sigma} - f_x$$

Total firm profit is domestic profit plus export profit across all $n$ foreign markets:

$$\pi(\varphi) = \pi_d(\varphi) + n\max\{0, \pi_x(\varphi)\}$$

#### Two Productivity Cutoffs

Since both $\pi_d$ and $\pi_x$ are increasing in $\varphi$, there are now two 
productivity thresholds:

$$\varphi^* \text{ such that } \pi_d(\varphi^*) = 0, \qquad 
\varphi^*_x \text{ such that } \pi_x(\varphi^*_x) = 0$$

The domestic cutoff $\varphi^*$ determines survival; the export cutoff $\varphi^*_x$ 
determines selection into exporting. For both non-exporters and exporters to coexist in equilibrium — a feature consistent with the empirical evidence that only a minority of firms export — we require $\varphi^*_x > \varphi^*$. From the zero-profit conditions:

$$\frac{r_x(\varphi^*_x)}{r_d(\varphi^*)} = \frac{f_x}{f} \quad \Longrightarrow \quad 
\varphi^*_x = \varphi^*\tau\left(\frac{f_x}{f}\right)^{\frac{1}{\sigma-1}}$$

This confirms that $\varphi^*_x > \varphi^*$ if and only if $\tau^{\sigma-1}f_x > f$: 
exporting requires a productivity advantage sufficient to overcome both the variable cost disadvantage (iceberg costs) and the fixed cost disadvantage (export entry cost). This condition generates the model's central prediction about the extensive margin of trade: **only the most productive firms export**.

The three-way sorting of firms that emerges — exit, domestic production only, export — is illustrated schematically as:

$$\underbrace{[0, \varphi^*)}_{\text{exit}} \quad 
\underbrace{[\varphi^*, \varphi^*_x)}_{\text{domestic only}} \quad 
\underbrace{[\varphi^*_x, \infty)}_{\text{export}}$$

#### Open Economy Aggregation

The open economy requires extending the aggregation to account for both domestic and 
foreign firms selling in each market. The total mass of firms selling in a given country is $M_T = M + nM_x$, where $M$ is the mass of domestic producers and $M_x$ is the mass of exporters from each of the $n$ foreign countries. The average productivity across all firms active in a market — domestic and foreign — is:

$$\tilde{\varphi}_T = \left[\frac{1}{M_T}\left(M\tilde{\varphi}^{\sigma-1} + 
nM_x\left(\frac{\tilde{\varphi}_x}{\tau}\right)^{\sigma-1}\right)\right]^{\frac{1}{\sigma-1}}$$

where $\tilde{\varphi}$ is average productivity among domestic producers and 
$\tilde{\varphi}_x$ is average productivity among domestic exporters:

$$\tilde{\varphi} = \left[\frac{1}{1-G(\varphi^*)}\int_{\varphi^*}^{\infty} 
\varphi^{\sigma-1}g(\varphi)\,d\varphi\right]^{\frac{1}{\sigma-1}}, \qquad 
\tilde{\varphi}_x = \left[\frac{1}{1-G(\varphi^*_x)}\int_{\varphi^*_x}^{\infty} 
\varphi^{\sigma-1}g(\varphi)\,d\varphi\right]^{\frac{1}{\sigma-1}}$$

Note that $\tilde{\varphi}_x > \tilde{\varphi}$ since exporters are drawn from a more 
productive subset of the firm distribution. The aggregate price index, revenue, profit, and output in the open economy take the same form as in the closed economy with $\tilde{\varphi}_T$ replacing $\tilde{\varphi}$ and $M_T$ replacing $M$:

$$P = \frac{M_T^{\frac{1}{1-\sigma}}}{\rho\tilde{\varphi}_T}, \qquad 
R = M_T r(\tilde{\varphi}_T), \qquad \Pi = M_T\pi(\tilde{\varphi}_T), \qquad 
Q = M_T^{\frac{\sigma}{\sigma-1}}q(\tilde{\varphi}_T)$$

#### Open Economy Equilibrium: Modified ZCP and FE

The free entry condition in the open economy accounts for the possibility of exporting. Denoting $p_x = (1-G(\varphi^*_x))/(1-G(\varphi^*))$ as the probability of being productive enough to export conditional on successful entry, average profit is:

$$\bar{\pi} = \pi_d(\tilde{\varphi}) + np_x\pi_x(\tilde{\varphi}_x) 
= f\left[\left(\frac{\tilde{\varphi}(\varphi^*)}{\varphi^*}\right)^{\sigma-1} - 1\right] 
+ np_x f_x\left[\left(\frac{\tilde{\varphi}_x(\varphi^*)}{\varphi^*_x(\varphi^*)}\right)^{\sigma-1} 
- 1\right]$$

This is the open economy ZCP condition: average profit now has two components — the 
contribution from domestic sales and the contribution from export sales, weighted by the probability of being an exporter. The free entry condition retains its closed-economy form:

$$\bar{\pi} = \frac{\delta f_e}{1 - G(\varphi^*)}$$

The open economy equilibrium is again found at the intersection of these two loci:

#### The Selection Effect of Trade

The key comparative static result is that the open economy equilibrium features a strictly higher productivity cutoff than autarky: $\varphi^*_{open} > \varphi^*_{autarky}$. The intuition can be decomposed by firm type:

- **Exporters** face two opposing forces: export opportunities raise profits, but the 
  entry of foreign competitors in the domestic market erodes domestic profits. For 
  exporters, the two effects partially offset but the net impact on profits is ambiguous.
- **Non-exporters** face only the second force: foreign entry erodes their domestic market share with no compensating export revenue. For non-exporters, trade unambiguously reduces profits, making it harder to survive.

The result is that opening to trade raises $\varphi^*$, forcing the least productive firms out of the market. The distribution of active firms shifts upward — average productivity rises — as low-productivity firms exit and are replaced by high-productivity firms that can profitably export. This reallocation of market share from less to more productive firms is the **selection effect** of trade, absent by construction from the symmetric Krugman model.

#### Welfare Effects of Trade

Welfare is summarized by the price index $P = M_T^{1/(1-\sigma)}/\rho\tilde{\varphi}_T$, or equivalently:

$$P^{1-\sigma} = M_T(\rho\tilde{\varphi}_T)^{\sigma-1}$$

Welfare is higher when $P$ is lower, which requires either more varieties ($M_T$ larger) or higher average productivity ($\tilde{\varphi}_T$ larger). Trade affects both:

- **Variety effect**: provided iceberg costs $\tau$ are not prohibitively high, the total   mass of varieties available to consumers rises — domestic varieties plus foreign imports — so $M_T > M_a$.
- **Productivity effect**: the selection mechanism ensures $\tilde{\varphi}_T > 
  \tilde{\varphi}_a$ always, since trade truncates the left tail of the active firm 
  distribution.

Melitz (2003) establishes a stronger result: **welfare in the open economy is always 
higher than in autarky, even if the total number of available varieties falls.** When 
trade costs are high, the variety effect may be weak or even negative (if many domestic firms exit). But the productivity effect is always present and always positive — the selection of more efficient firms into production reduces the price index even if the range of available varieties narrows. This is the Melitz model's key welfare result and its most important departure from Krugman: gains from trade do not require an expansion of variety; they can arise purely from the reallocation of resources toward more productive firms.

### Conclusion

The Melitz model provides a unified framework for understanding how trade affects not just the volume of production but the composition of firms within an industry. Its main contributions and predictions are:

- **Firm heterogeneity and selection**: firms differ in productivity, and trade 
  systematically selects the most productive into exporting while forcing the least 
  productive to exit. This within-industry reallocation raises average productivity even without any improvement at the level of individual firms.
- **Two margins of trade**: trade operates on both the intensive margin (existing exporters sell more) and the extensive margin (new firms begin exporting). The model predicts that   exporters are larger, more productive, and more profitable than non-exporters within the same industry — consistent with the empirical evidence of Bernard and Jensen (1999) and Trefler (2004).
- **Welfare gains beyond variety**: unlike Krugman, where welfare gains arise solely from an expansion of variety, the Melitz model identifies a productivity channel that operates even when the variety effect is weak. The two channels together provide a richer account of the gains from trade.
- **Implications for trade policy**: because trade induces exit among low-productivity 
  firms, protectionist policies that shelter these firms from competition prevent the 
  reallocation that generates aggregate productivity gains. The model formalizes the idea that protecting incumbents has a cost beyond the standard deadweight loss — it also suppresses the dynamic gains from firm turnover.

The Melitz framework has become the workhorse model of the modern trade literature. Its combination of tractability, empirical discipline, and rich predictions about firm-level behavior makes it the natural starting point for any analysis of trade policy, firm dynamics, or the interaction between globalization and productivity growth.
