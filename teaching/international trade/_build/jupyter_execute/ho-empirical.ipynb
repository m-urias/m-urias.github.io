# Empirical Tests of the Heckscher-Ohlin Theory

## The Leontief Paradox

The HO model's central prediction is that countries export goods that use their abundant factors intensively. For the United States in the postwar period, this prediction seemed straightforward to test: the US was widely regarded as the most capital-abundant economy in the world, and the model therefore predicted that US exports should be capital-intensive relative to US imports.

The first systematic empirical test was conducted by Wassily Leontief, published in 1953, using input-output tables for the US economy. Input-output analysis allows one to trace the full chain of factor requirements embodied in a unit of output — not just the direct labor and capital used in final assembly, but also the indirect factor content embedded in intermediate inputs. Leontief used this methodology to compute the capital and labor embodied in a representative bundle of US exports and a representative bundle of US imports.

The result was startling: the capital-labor ratio embodied in US *imports* exceeded that embodied in US *exports*. The most capital-abundant country in the world was apparently exporting labor-intensive goods and importing capital-intensive ones. This finding — the direct opposite of the HO prediction — became known as the **Leontief paradox**, and it generated a large literature attempting to explain, qualify, or overturn it.

### Proposed Resolutions

Several explanations were advanced for why Leontief's results might not, in fact, refute the HO model:

- **Technology differences**: the HO model assumes identical production technologies across countries, but if US workers are more productive than foreign workers — due to superior technology or organizational practices — then the effective factor endowment of the US in efficiency units may be less capital-intensive than raw capital stock figures suggest.
- **Human capital**: Leontief's study measured only physical capital and raw labor, ignoring the fact that US workers were far better educated and trained than workers elsewhere. Once  human capital is included in the factor endowment, the US may in fact be *labor*-abundant  in the relevant sense.
- **Natural resources**: the US imports many resource-intensive goods — oil, raw materials, agricultural products — whose production happens to be capital-intensive. Omitting natural resources as a factor biases the results.
- **Trade policy**: the HO model is a model of free trade equilibria, but the US was not engaged in free trade. Tariffs and quotas may have distorted the pattern of trade away from the model's predictions.

These explanations have merit, but there is a more fundamental issue: Leontief was not 
actually testing the correct implication of the HO model.

### The Correct Test: Leamer (1980)

The appropriate test of the HO model was formalized by Edward Leamer (1980), who showed that Leontief's comparison of capital-labor ratios in exports and imports was logically flawed in the presence of a trade imbalance. The correct prediction of the HO model is not about the relative factor intensity of exports versus imports, but about the factor content of *net* trade relative to factor endowments.

To state this precisely, consider a world with many countries $i = 1, \ldots, C$, many goods $j = 1, \ldots, N$, and many factors $k = 1, \ldots, M$. Let $A$ be the $(M \times N)$ matrix of factor requirements, where element $a_{kj}$ denotes the amount of factor $k$ required to produce one unit of good $j$. For the two-factor case:

$$
A = \begin{pmatrix} a_{1L} & a_{2L} \\ a_{1K} & a_{2K} \end{pmatrix}
$$

Let $\mathbf{T}^i$ be the $(N \times 1)$ vector of net exports of country $i$ (exports minus imports, by good). Then the **factor content of net trade** for country $i$ is:

$$F^i = A \mathbf{T}^i$$

where $F^i_k$ measures how much of factor $k$ is embodied in country $i$'s net exports. The HO model predicts that $F^i_k > 0$ — a country is a net exporter of the services of factor $k$ — if and only if country $i$ is relatively abundant in factor $k$. This is the correct statement to test empirically, and it differs from Leontief's comparison in a subtle but important way: Leontief compared ratios *within* a trade flow, whereas the correct test concerns the *net* factor content relative to the country's endowment.

### The Heckscher-Ohlin-Vanek Theorem

The two-country, two-factor HO model delivers a prediction about the *direction* of trade in 
goods. With many countries, goods, and factors, the natural generalization shifts the 
prediction to the *factor content* of trade. This generalization is due to Vanek (1968) and 
forms the basis for the main empirical tests of the HO framework.

#### Setup

Let $Y^i$ be the $(N \times 1)$ vector of outputs across industries for country $i$, and let 
$D^i$ be the corresponding vector of demands. The vector of net exports is $T^i = Y^i - D^i$. 
The **factor content of trade** is then:

$$F^i \equiv A(Y^i - D^i) = AT^i$$

where $A$ is the $(M \times N)$ matrix of factor requirements. The $k$-th element $F^i_k$ 
measures the net export of factor $k$'s services embedded in country $i$'s trade: $F^i_k > 0$ 
means factor $k$ is implicitly exported; $F^i_k < 0$ means it is implicitly imported.

#### Derivation

The HOV prediction follows from three assumptions: (i) full employment, (ii) identical and 
homothetic preferences, and (iii) balanced trade.

Full employment requires that all factor endowments are absorbed in production:

$$V^i = AY^i$$

Identical and homothetic preferences with balanced trade imply that each country's consumption 
is proportional to its share $s^i$ of world GDP:

$$AD^i = s^i AD^w = s^i AY^w = s^i V^w$$

where $s^i = \text{GDP}^i / \text{GDP}^w$. Substituting into the definition of $F^i$:

$$F^i = AY^i - AD^i = V^i - s^i V^w$$
```{admonition} Heckscher-Ohlin-Vanek (HOV) Theorem
:class: definition
The factor content of trade in factor $k$ satisfies $F^i_k = V^i_k - s^i V^w_k$. 
In particular, $F^i_k > 0$ (country $i$ is a net exporter of factor $k$) if and only if 
factor $k$ is abundant in the sense that $V^i_k / V^w_k > s^i$.
```

#### Resolving the Leontief Paradox

The HOV framework clarifies why Leontief's test was logically flawed. Leontief compared the 
capital-labor *ratio* in export goods against that in import goods — a comparison of 
intensities, not quantities. The HOV theorem makes a prediction about the *net* factor 
content of trade, which depends on both the intensity and the scale of trade flows.

In the two-factor case, the HOV predictions for capital and labor are:

$$F_K^i = K^i - s^i K^w, \qquad F_L^i = L^i - s^i L^w$$

If the US is capital-abundant relative to labor — meaning $K^{US}/K^w > L^{US}/L^w$ — then 
the correct HOV prediction is:

$$\frac{K^{US}}{L^{US}} > \frac{K^{US} - F_K^{US}}{L^{US} - F_L^{US}}$$

That is, the capital-labor ratio in US *production* should exceed the capital-labor ratio in 
US *consumption* — a statement about net factor flows, not about the intensity of any 
particular export or import bundle.

**Proof.** From the HOV conditions and market clearing:

$$K^w = \frac{K^i - F_K^i}{s^i}, \qquad L^w = \frac{L^i - F_L^i}{s^i}$$

Therefore:

$$\frac{K^i}{K^w} = s^i \frac{K^i}{K^i - F_K^i}, \qquad \frac{L^i}{L^w} = s^i \frac{L^i}{L^i - F_L^i}$$

Capital abundance ($K^i/K^w > L^i/L^w$) then implies:

$$\frac{K^i}{K^i - F_K^i} > \frac{L^i}{L^i - F_L^i}$$

which is precisely the production-consumption comparison above. $\square$

#### A Numerical Illustration

To make the distinction between Leontief's method and the HOV approach concrete, consider 
the following example. Suppose the US economy produces two goods — machinery (M) and 
textiles (T) — using capital and labor with the following input requirements per \$1M of 
output:

| Good | Capital | Labor | $K/L$ |
|---|---|---|---|
| Machinery (exported) | 300 | 100 | 3.0 |
| Textiles (imported) | 600 | 100 | 6.0 |

**Leontief's approach** compares the $K/L$ ratio of the export bundle to that of the import 
bundle. Machinery has $K/L = 3.0$ and textiles have $K/L = 6.0$, so the US appears to be 
exporting the labor-intensive good and importing the capital-intensive good — the Leontief 
paradox.

**The HOV approach** instead computes the net factor content of trade. Suppose the US exports 
\$2M of machinery and imports \$1M of textiles:

$$T = \begin{bmatrix} 2 \\ -1 \end{bmatrix}, \quad 
A = \begin{bmatrix} 300 & 600 \\ 100 & 100 \end{bmatrix}, \quad
AT = \begin{bmatrix} (2)(300) - (1)(600) \\ (2)(100) - (1)(100) \end{bmatrix} 
= \begin{bmatrix} 0 \\ 100 \end{bmatrix}$$

On net, the US is exporting 100 units of labor and zero units of capital — seemingly 
consistent with Leontief's finding. But now suppose the volume of trade is larger: \$10M of 
machinery exported and \$2M of textiles imported:

$$T = \begin{bmatrix} 10 \\ -2 \end{bmatrix}, \quad
AT = \begin{bmatrix} (10)(300) - (2)(600) \\ (10)(100) - (2)(100) \end{bmatrix}
= \begin{bmatrix} 1800 \\ 800 \end{bmatrix}$$

Now the US is a net exporter of both capital and labor, with a capital-to-labor ratio of 
$1800/800 = 2.25$ in its net trade. The $K/L$ ratio of the net export bundle need not equal 
the $K/L$ ratio of any individual good. Leontief's error was to draw inferences about factor 
abundance from good-level intensity comparisons, while ignoring the scale of trade. The HOV 
approach, by focusing on net factor content, corrects this.

### Empirical Performance of the HOV Model

The HOV theorem provides a clean empirical test. The prediction is:

$$AT^i = V^i - s^i V^w$$

The left-hand side can be computed from data on trade flows $T^i$ and technology $A$; the 
right-hand side from data on factor endowments. The question is how well the two sides match.

#### Bowen, Leamer, and Sveikauskas (1987)

The first comprehensive test of the HOV theorem was conducted by Bowen, Leamer, and 
Sveikauskas (1987), who extended Leontief's analysis to 27 countries and 12 factors of 
production. For each country-factor pair, they tested whether the sign of the measured factor 
content of trade matched the HOV prediction. The sign test passed in only 61 percent of cases 
— not much better than random assignment — and the Leontief paradox remained alive in the 
broader cross-country data.

#### Trefler (1995)

Trefler (1995) pushed the analysis further using data on 33 countries and 9 factors, testing 
both the direction and the magnitude of the HOV predictions. The results were damaging: the 
sign test was consistent with the HOV prediction in only 50 percent of cases, and the rank 
test in 60 percent. Trefler also documented a striking pattern he termed the **mystery of the 
missing trade**: observed factor trade flows were far smaller in magnitude than the HOV model 
predicts. Trefler's diagnostic analysis pointed to the assumption of identical technologies 
across countries as the primary source of the model's failure — a finding consistent with 
the resolution of the Leontief paradox discussed above.

#### Davis and Weinstein (2001)

The most influential rehabilitation of the HOV framework was provided by Davis and Weinstein 
(2001), who showed that the model's empirical failures were largely attributable to its most 
restrictive auxiliary assumptions rather than to the core factor-endowments mechanism. When 
they relaxed the assumptions of identical technologies, a common set of goods produced across 
all countries, and costless trade, the HOV predictions improved dramatically:

<figure>
    <img src="_static/figs_ho/ho_13.png"
         alt="Davis and Weinstein results"
         width="600"
         style="display: block; margin-left: auto; margin-right: auto;">
    <figcaption style="font-style: italic; text-align: left;">
    Davis and Weinstein (2001): measured factor content of trade against HOV predictions, 
    under progressively relaxed assumptions. Each step away from the strict HOV assumptions 
    improves the fit substantially.
    </figcaption>
</figure>

#### Assessment

The empirical literature on HOV delivers a nuanced verdict. The strict version of the 
theorem — with identical technologies, costless trade, and full diversification — performs 
poorly and the Leontief paradox is a genuine anomaly within that framework. But the 
anomaly is largely resolved once the theorem's most counterfactual assumptions are relaxed. 
Factor endowments do appear to shape the factor content of trade in the manner the HO 
mechanism predicts; the difficulty lies in the auxiliary assumptions required to derive the 
clean HOV prediction rather than in the core economic logic.

Whether the amended model — with technology differences, trade costs, and incomplete 
specialization layered on top — should still be called the HO model is a matter of 
interpretation. What is clear is that factor endowments remain an empirically relevant 
determinant of trade patterns, even if the strict HOV theorem overstates the precision with 
which endowments alone can account for observed trade flows.
