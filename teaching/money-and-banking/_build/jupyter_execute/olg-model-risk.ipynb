# OLG Model with a Risky Asset

According to the rate of return dominance principle, if assets are identical in terms of their liquidity and risk then agents should only hold the asset with the higher rate of return.  An implication of this principle is that multiple assets are held in equilibrium if and only if they generate the same rate of return.  

We now break the assumption that assets are equivalent in terms of risk.  This allows both assets to be held in equilibrium even when the risky asset pays a higher *expected rate of return* compared to fiat currency.

## Portfolio Choice with a Risky Asset

Suppose there is a risky asset whose real returns $R$ are a random variable with mean $\mu$ and variance $\sigma^2$.  An young agent who purchases $a$ units of the asset enjoys consumption $aR$ when old.  This is in contrast to fiat currency which pays a rate of return $\nu_{t+1}/\nu_t$ with certainty.

To highlight the role of risk in determining the optimal portfolio we assume, without loss of generality, that agents only value consumption when old so that lifetime utility is given by

$$
U = u(c_{2,t+1})
$$

Agents must decide how much of their wealth $y$ to invest in fiat currency and the risky asset to maximize their *expected utility*.  The problem is given by

$$
\begin{align}
\max &\hspace{2mm} \mathbb{E}[u(c_{2,t+1})] \\
\text{s.t.} \quad &\nu_t m_t+a_t \leq y \\
&c_{2,t+1}=\nu_{t+1}m_t+a_tR_t
\end{align}
$$

where we set $c_{1,t}=0$ in the budget constraint of the young since agents do not value consumption when young.  The problem is equivalently expressed as

$$
\begin{align}
\max &\mathbb{E}[u(q_t(\nu_{t+1}/\nu_t) +a_tR_t)] \\
\text{s.t.} \quad & q_t+a_t \leq y
\end{align}
$$

Agents are simply maximizing the expected utility of their portfolio subject to a wealth constraint.  The problem agents face is that the rate of return $R$ is uncertain.  Moreover, so long as agents are risk averse we have that $\mathbb{E}[u(c)]<u(\mathbb{E}[c])$ so that agents don't only care about expected returns, but also the degree of uncertainty over those returns.  We can formally disentangle the role of expected return and risk in the following way.

Let $X_t \equiv q_t(\nu_{t+1}/\nu_t) +a_t\mathbb{E}[R_t]$ be the expected return on the portfolio and let $\epsilon \equiv a_t(R_t-\mathbb{E}_t[R])$ denote how much the realized rate of return differs from its expected value.  Of course we have that $\mathbb{E}[\epsilon]=0$ so it is a zero-mean risk component of the asset.  The household objective can be written as

$$
\mathbb{E}[u(X_t+\epsilon_t)]
$$

We now consider a second-order Taylor expansion around $R=\mathbb{E}[R]$, or $\epsilon=0$:

$$
\mathbb{E}[u(X_t+\epsilon_t)] = u(X_t)+u'(X_t)\underbrace{\mathbb{E}[\epsilon]}_{0}+\frac{1}{2}u''(X_t)\underbrace{\mathbb{E}[\epsilon^2]]}_{a^2Var[R]}
$$

Notice that the second term in the Taylor expansion reduces to zero while the third term is precisely the variance of $R$.  We therefore have that

$$
\mathbb{E}[u(c_{2,t+1})]=u(X_t)+\frac{1}{2}u''(X_t)a_t^2 Var(R_t), \quad X_t=q_t(\nu_{t+1}/\nu_t) +a_t\mathbb{E}[R_t]
$$

This decomposes the expected utility of a portfolio into a component driven by expected returns ($X$) and a component which captures risk ($Var(R)$).  The problem of the individual is therefore:

$$
\begin{align}
\max \hspace{2mm} &u(X_t)+\frac{1}{2}u''(X_t)a_t^2 Var(R_t) \\
\text{s.t.} \quad  X_t &=q_t(\nu_{t+1}/\nu_t) +a_t\mathbb{E}[R_t] \\
q_t+a_t &\leq y
\end{align}
$$

## An Example with Log Utility

In order to provide a concrete example, let's use log utility $u(c)=\log(c)$.  In this case the problem of a household is given by

$$
\begin{align}
\max \hspace{2mm} &\log(X_t) - \frac{1}{2}\frac{a^2 Var(R)}{X^2} \\
\text{s.t.} \quad  X_t &=q_t(\nu_{t+1}/\nu_t) +a_t\mathbb{E}[R_t] \\
q_t+a_t &\leq y
\end{align}
$$

Simplify the problem by substituting the income constraint into the definition of the portfolio to say $X=(y-a_t)(\nu_{t+1}/\nu_t)+a_t\mathbb{E}[R_t]$ and define $\mu=\mathbb{E}[R]-\nu_{t+1}/\nu_t$ as the excess return of the risky asset over money.  These simplifications allow us to write the problem as

$$
\begin{align}
\max_{a} \hspace{2mm} \log \left( (\nu_{t+1}/\nu_t)y+\mu a \right)-\frac{1}{2}\frac{a^2 Var(R)}{\left( (\nu_{t+1}/\nu_t)y+\mu a\right)^2}
\end{align}
$$

The first-order condition to this problem is

$$
\frac{\mu}{X}=\frac{a Var(R)}{X^2}-\frac{a^2 Var(R)\mu}{X^3}
$$

As a decent approximation, we drop the final term to say

$$
a = \frac{\mu}{Var(R)}X
$$

The numerator captures the fact that higher excess returns encourage the agent to allocate more of their portfolio into the risky asset, but the denominator shows that more risk discourages investment in the risky asset.

Substituting out $X$ gives us a closed form expression for optimal investment in the risky asset:

$$
a = \frac{\mu}{\sigma^2}\left(q_t(\nu_{t+1}/\nu_t) +a_t\mathbb{E}[R_t]\right)
$$

which simplifies to 

$$
a = \left( \mathbb{E}[R]-\mu \right) \left( \frac{\mu y}{\sigma^2-\mu^2}\right)
$$

and therefore amount of real money demand is simply

$$
\nu_t m_t = y-a_t
$$

It is sufficient that $\sigma^2<\mu^2$ for $\partial a/\partial \mu>0$.  That is, so long as the risky asset has a standard deviation lower than the expected excess return, agents invest more in the risky when it's excess returns rise.  

We now have a link between money creation and risk taking in the economy.  If the government increases the growth rate of money supply, this lowers the rate of return on money ($\downarrow \nu_{t+1}/\nu_t$) and thereby increases the excess returns from the risky asset ($\uparrow \mu$).  Agents invest more in the risky asset and less in currency.  Similar logic suggests a link between the fundamentals underlying the risky asset and the value of money.  If the average returns of the risky asset fall or the standard deviation of the asset rises, individuals rellocate their portfolio away from the risky asset and toward fiat currency.  This implying a higher demand for fiat currency and hence a higher equilibrium value.

## Endogenizing the Rate of the Return on the Risky Asset

The analysis thus far has taken the rate of return on the risky asset as given.  In reality, the rate of return on risky assets (e.g. equities) are determined by the price they trade at which equates the demand for that asset to it's supply.  Consider the risky asset as an equity share in a firm.  The share entitles the holder to a stream of dividends $d$ which are generated by the profits of the firm.  This stream of dividends is uncertain though, because firms are constantly hit by random shocks which generates random profits.  The holder of the share can sell the share in the secondary market at price $p^e$.  The rate of return on the share is

$$
R_{t+1}^e = \frac{p^e_{t+1}+d_{t+1}}{p_t^e} 
$$

The price that the asset sells for in the secondary market is determined in equilibrium by equating the supply of that asset to it's demand.  Hence, the rate of return $R^e$ is endogenous.

For clarity, we will continue using log perferences so that the problem facing an individual is given by

$$
\begin{align}
\max &\mathbb{E}[\log c_{2,t+1}] \\
\text{s.t.} \quad &\nu_t m_t + p^e_t a_t = y \\
& c_{2,t+1} = \nu_{t+1}m_t+(p^e_{t+1}+d_{t+1})a_t
\end{align}
$$
